```python
import pickle

with open("../tokenization/vocab.pkl", "rb") as f:
    vocab = pickle.load(f)

# vocab: dict[int, bytes]
items = sorted(
    vocab.items(),
    key=lambda item: len(item[1]),
    reverse=True,
)

print("Top 50 longest vocab tokens:")
for token_id, token_bytes in items[:50]:
    try:
        decoded = token_bytes.decode("utf-8")
    except UnicodeDecodeError:
        decoded = "<not valid utf-8>"

    print(
        f"id={token_id:5d} "
        f"len={len(token_bytes):4d} "
        f"bytes={repr(token_bytes)} "
        f"text={repr(decoded)}"
    )
```
